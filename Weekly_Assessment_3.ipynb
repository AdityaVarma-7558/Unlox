```python
# 1. Memory-usage difference between list and tuple
import sys
my_list = [1, 2, 3, 4, 5]
my_tuple = (1, 2, 3, 4, 5)
print(sys.getsizeof(my_list))
print(sys.getsizeof(my_tuple))
```

```python
# 2. Swap two variables in a single line
a = 10
b = 20
a, b = b, a
print(a, b)
```

```python
# 3. Find the data type of every element in a mixed list
mixed_list = [42, "AI", 3.14, True, None, [1, 2]]
for item in mixed_list:
    print(type(item))
```

```python
# 4. Check if a string is a palindrome
def is_palindrome(text):
    clean_text = text.replace(" ", "").lower()
    return clean_text == clean_text[::-1]
test_string = "racecar"
print(is_palindrome(test_string))
```

```python
# 5. Difference between / and // operators
a = 10
b = 3
print(a / b)
print(a // b)
```

```python
# 6. Extract the last 3 characters of a string using slicing
my_string = "ArtificialIntelligence"
last_three = my_string[-3:]
print(last_three)
```

```python
# 7. Count the number of vowels in a user-provided sentence
user_sentence = input("Enter a sentence: ")
vowels = "aeiouAEIOU"
vowel_count = sum(1 for char in user_sentence if char in vowels)
print(vowel_count)
```

```python
# 8. "f-strings" for variable interpolation
model_name = "Neural Network"
accuracy = 98.5
output = f"The {model_name} achieved an accuracy of {accuracy}%."
print(output)
```

```python
# 9. Create a list of numbers from 1 to 20 and print only the even ones
numbers = list(range(1, 21))
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
# 10. Difference between range(5) and list(range(5))
r_obj = range(5)
l_obj = list(range(5))
print(type(r_obj), r_obj)
print(type(l_obj), l_obj)
```
