```python
# 1. Write a recursive function for the Fibonacci sequence.
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(5):
    print(fibonacci(i))
```

```python
# 2. Explain "Memoization" and how it helps recursive functions.
"""Memoization is an optimization technique used to speed up programs by storing the results of expensive function calls in a cache (like a dictionary) and returning the cached result when the exact same inputs occur again. In recursive functions, it prevents the algorithm from redundantly calculating the same branches over and over, which drastically reduces the time complexity."""
```

```python
# 3. Use reduce() from functools to find the product of a list.
from functools import reduce

def multiply_list(numbers):
    return reduce(lambda x, y: x * y, numbers)

print(multiply_list([1, 2, 3, 4, 5]))
```

```python
# 4. Write a function to check if a number is an "Armstrong Number.".
def is_armstrong(number):
    num_str = str(number)
    power = len(num_str)
    total = sum(int(digit) ** power for digit in num_str)
    return total == number

print(is_armstrong(153))
```

```python
# 5. Create a closure that remembers a base value and adds it to an input.
def create_adder(base_value):
    def adder(input_value):
        return base_value + input_value
    return adder

add_five = create_adder(5)
print(add_five(10))
```

# 6. Explain the "LEGB" rule in Python scope.
The LEGB rule dictates the order in which Python searches for variable names to resolve their scope. It searches in this specific sequence and stops at the first match:

Local (L): Names assigned within the current function.

Enclosing (E): Names in the local scope of any enclosing functions, from the innermost to the outermost.

Global (G): Names assigned at the top-level of the module file or explicitly declared as global.

Built-in (B): Names preassigned in Python's built-in namespace (e.g., len, range).

```python
# 7. Write a function that converts a decimal number to binary without bin().
def decimal_to_binary(n):
    if n == 0:
        return "0"
    binary_str = ""
    while n > 0:
        binary_str = str(n % 2) + binary_str
        n = n // 2
    return binary_str

print(decimal_to_binary(10))
```

```python
# 8. Create a "Generator" function using yield to produce infinite even numbers.
def infinite_evens():
    num = 0
    while True:
        yield num
        num += 2

evens = infinite_evens()
print(next(evens))
print(next(evens))
print(next(evens))
```

```python
# 9. How do you pass a function as an argument to another function?
def apply_function(func, value):
    return func(value)

def square(x):
    return x * x

print(apply_function(square, 5))
```

```python
# 10. Write a function to validate a password (min 8 chars, 1 digit, 1 symbol).
def validate_password(password):
    if len(password) < 8:
        return False

    has_digit = any(char.isdigit() for char in password)
    has_symbol = any(char in "!@#$%^&*()_+-=[]{}|;':\",./<>?" for char in password)

    return has_digit and has_symbol

print(validate_password("Password123!"))
```
