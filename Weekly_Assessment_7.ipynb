```python
# 1. Define a class Car with attributes and a method to display details.
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_details(self):
        print(f"Car Details: {self.year} {self.make} {self.model}")
```

# 2. Explain the purpose of the _init_method.
The __init__ method is a constructor in Python. It is automatically
called when a new instance of a class is created to initialize
the object's attributes with starting values.

```python
# 3. Create a subclass ElectricCar that inherits from Car.
class ElectricCar(Car):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity
```

```python
# 4. Demonstrate "Method Overriding" in inheritance.
class ElectricCar(Car):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity

    def display_details(self):
        print(f"Electric Car Details: {self.year} {self.make} {self.model} with a {self.battery_capacity}kWh battery")
```

```python
# 5. Explain "Encapsulation" by creating private attributes using
"""
Encapsulation bundles data and methods within a class while restricting
direct access to some components. In Python, attributes are made private
by prefixing them with a double underscore (__).
"""
class EncapsulatedCar:
    def __init__(self, make, model):
        self.make = make
        self.__model = model
```

```python
# 6. Use @property to create getter and setter methods.
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative.")
```

```python
# 7. What is "Multiple Inheritance"? Provide a small code structure.
"""
Multiple inheritance is a feature where a single class can inherit
attributes and methods from more than one parent class.
"""
class Engine:
    def start_engine(self):
        return "Engine is roaring."

class EntertainmentSystem:
    def play_music(self):
        return "Playing radio."

class AdvancedCar(Engine, EntertainmentSystem):
    pass
```

```python
# 8. Create a class method using the @classmethod decorator.
class Company:
    company_name = "Tech Solutions Inc."

    @classmethod
    def get_company_name(cls):
        return f"The company is {cls.company_name}"
```

# 9. Explain the difference between a class method and a static method.
A class method takes the class itself as its first
argument and can access or modify class state.

A static method takes neither the instance nor the class as its first argument and behaves like a normal function within the class namespace.


```python
# 10. Write a try-except-finally block to handle a ZeroDivisionError.
def divide(a, b):
    try:
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error: You cannot divide a number by zero.")
    finally:
        print("Execution of the division function is complete.\n")
```
