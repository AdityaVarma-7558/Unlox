**4. Explain the difference between is and ==.**

The == operator compares the actual values of two objects to check if they are equivalent. The is operator compares the memory addresses of the two objects to check if they are the exact same object in memory.

```python
9# 5. Create a simple calculator using if-elif-else logic.
num1 = float(input())
operator = input()
num2 = float(input())

if operator == '+':
    print(num1 + num2)
elif operator == '-':
    print(num1 - num2)
elif operator == '*':
    print(num1 * num2)
elif operator == '/':
    print(num1 / num2)
else:
    print("Invalid operator")
```

```python
# 6. Write a program to find the largest number in a list without using max().
numbers = [3, 14, 1, 5, 9, 2]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)
```

```python
# 9. Write a script that prints the multiplication table of a given number.
num = int(input())
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

**10. Explain the use of the in keyword in Python.**

The in keyword is used as a membership operator to verify if a specific value exists inside an iterater. It evaluates to True if the value is present and False if it is not. It is also used to iterate through items sequentially when writing a for loop.
